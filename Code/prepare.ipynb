```python
data_df = [
    'Year_of_Release',
    'Critic_Count',
    'User_Count'
]

df[data_df].isna().sum()

df[data_df] = df[data_df].astype('Int64')

df
```

```python
df.isna().sum()
```

```python
df["User_Score"].value_counts(dropna=False).head(20)

df["User_Score"] = df["User_Score"].replace("tbd", pd.NA)
df["User_Score"].isna().sum()
```

```python
df["User_Score"] = pd.to_numeric(df["User_Score"], errors="coerce")

df.info()
```

```python
categories = [
    'Genre',
    'Publisher',
    'Developer',
    'Rating'
]

df[categories] = df[categories].fillna('Unknown')

df
```

```python

rating_map = {
    "E": "Everyone",
    "M": "Mature",
    "T": "Teen",
    "E10+": "Everyone 10+",
    "K-A": "Kids to Adults",
    "AO": "Adults Only",
    "EC": "Early Childhood",
    "RP": "Rating Pending",
    "Unknown": "Unknown"
}

df["Rating"] = df["Rating"].replace(rating_map)

df
```
